# A2UI JavaScript kernel smoke test

This tests the same form as [smoke_test.ipynb](./smoke_test.ipynb). Run the form cell, enter a name, and click **Submit**. The callback replaces the form with a confirmation. Run the last cell to inspect the action.

```javascript
const { A2UISurface } = await import('/files/a2ui/kernel.js');
globalThis.surface = new A2UISurface();
globalThis.actionData = undefined;

const surfaceId = 'manual-smoke-test';
const catalogId = 'https://a2ui.org/specification/v1_0/catalogs/basic/catalog.json';
surface.send([
  {version: '1.0', createSurface: {surfaceId, catalogId}},
  {version: '1.0', updateDataModel: {surfaceId, path: '/person/name', value: ''}},
  {version: '1.0', updateDataModel: {surfaceId, path: '/status', value: 'Waiting for submission'}},
  {version: '1.0', updateComponents: {surfaceId, components: [
    {id: 'root', component: 'Column', children: ['title', 'name', 'submit', 'status'], align: 'stretch'},
    {id: 'title', component: 'Text', text: 'A2UI Jupyter renderer smoke test'},
    {id: 'name', component: 'TextField', label: 'Name', placeholder: 'Enter a name', value: {'@path': '/person/name'}},
    {id: 'submit', component: 'Button', child: 'submit_label', variant: 'primary', action: {event: {name: 'submit', context: {name: {'@path': '/person/name'}}}}},
    {id: 'submit_label', component: 'Text', text: 'Submit'},
    {id: 'status', component: 'Text', text: {'@path': '/status'}}
  ]}}
]);

surface.onAction((action) => {
  globalThis.actionData = action;
  surface.send([
    {version: '1.0', updateDataModel: {
      surfaceId, path: '/status', value: `Submitted: ${action.action.context.name}`
    }},
    {version: '1.0', updateComponents: {surfaceId, components: [
      {id: 'root', component: 'Column', children: ['title', 'status'], align: 'stretch'}
    ]}}
  ]);
});

display(surface);
```
