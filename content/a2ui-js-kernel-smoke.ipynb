# A2UI JavaScript kernel adapter smoke test

Run the cells in order. Cell 2 shows a surface. Click **Go**, then run cell 3 to read the action. Run cell 4 to push an update into the surface.

```javascript
const adapterUrl = globalThis.document
  ? new URL('../files/a2ui/kernel.js', document.baseURI)
  : new URL('../../../../files/a2ui/kernel.js', location.href);
const { A2UISurface } = await import(adapterUrl.href);
globalThis.A2UISurface = A2UISurface;
'adapter loaded'
```

```javascript
const CATALOG = 'https://a2ui.org/specification/v1_0/catalogs/basic/catalog.json';
globalThis.surface = new A2UISurface();
globalThis.surface.send([
  {version: '1.0', createSurface: {surfaceId: 's1', catalogId: CATALOG}},
  {version: '1.0', updateComponents: {surfaceId: 's1', components: [
    {id: 'root', component: 'Column', children: ['title', 'go']},
    {id: 'title', component: 'Text', text: 'Hello from the JS kernel'},
    {id: 'go', component: 'Button', child: 'go_label', action: {event: {name: 'go'}}},
    {id: 'go_label', component: 'Text', text: 'Go'}
  ]}}
]);
display(globalThis.surface);
```

```javascript
JSON.stringify(globalThis.surface.takeAction())
```

```javascript
globalThis.surface.send({version: '1.0', updateComponents: {surfaceId: 's1', components: [
  {id: 'title', component: 'Text', text: 'Updated from the kernel'}
]}});
'update sent'
```
